```python
import nltk
import pandas as pd

# Download resources required for tokenization and POS tagging
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
with open("Article.txt", "r", encoding="utf-8") as file:
    article = file.read()

print(article)
```

```python
words = nltk.word_tokenize(article)

print("First 50 tokens:")
print(words[:50])
```

```python
pos_tags = nltk.pos_tag(words)

print("First 50 POS tags:")
print(pos_tags[:50])
```

```python
noun_tags = ['NN', 'NNS', 'NNP', 'NNPS']
verb_tags = ['VB', 'VBD', 'VBG', 'VBN', 'VBP', 'VBZ']

extracted_words = []

for word, tag in pos_tags:
    if tag in noun_tags or tag in verb_tags:
        extracted_words.append([word, tag])

print(extracted_words[:50])
```

```python
df = pd.DataFrame(extracted_words, columns=["Word", "POS_Tag"])

df.head(50)
```

```python
output_filename = "Kritan_POS_01.csv"

df.to_csv(output_filename, index=False)

print("CSV file created successfully!")
```
