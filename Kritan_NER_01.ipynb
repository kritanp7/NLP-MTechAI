```python
!python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")

print("spaCy model loaded successfully!")
```

```python
with open("Article.txt", "r", encoding="utf-8") as file:
    article = file.read()

print(article)
```

```python
doc = nlp(article)

print("Named Entities Found:\n")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python
entities = []

for ent in doc.ents:
    entities.append([ent.text, ent.label_])

print("Total entities found:", len(entities))
```

```python
df = pd.DataFrame(
    entities,
    columns=["Entity", "Entity_Type"]
)

df.head(20)
```

```python
output_filename = "Kritan_NER_01.csv"

df.to_csv(output_filename, index=False)

print("CSV file created successfully!")
print("File:", output_filename)
```

```python
entity_summary = df["Entity_Type"].value_counts()

print("Summary of Named Entity Types:")
print(entity_summary)
```

## Entity Type Summary

The spaCy NER model identified different types of named entities in the
news article.

Some common entity types are:

- PERSON: Names of people.
- ORG: Organizations, companies, institutions, or agencies.
- GPE: Countries, cities, states, and other geopolitical entities.
- DATE: Dates or periods of time.
- MONEY: Monetary values.
- CARDINAL: Numerical values that do not belong to another specific category.
- NORP: Nationalities, religious groups, or political groups.

The frequency table above shows the number of entities detected for each
category in the selected news article.
